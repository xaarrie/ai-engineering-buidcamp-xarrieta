Importamos la funcion rag
Initialize index descarga la documentación hace los chunks y crea un indice minisearch

```python
import rag
import importlib
print(rag.__file__)
importlib.reload(rag)
index = rag.initialize_index()
```

Creamos el cliente openai y creamos la instancia de RAG

```python
from openai import OpenAI
openai_client = OpenAI()
rag_instance = rag.RAG(
    index,
    openai_client,
    model_name='gpt-4o-mini'
)
```

y le hacemos una pregunta

```python
response = rag_instance.rag('llm as a judge')
print(response.answer)

print(f"Found answer: {response.found_answer}")
print(f"Confidence: {response.confidence}")
print(f"Answer type: {response.answer_type}")
print(f"Follow-up questions: {response.followup_questions}")
```
